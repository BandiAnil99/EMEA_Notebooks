Run the cells from top to bottom. Each section displays the results for its questions.

```python
import pandas as pd

data = {
    "Customer_ID": [101,102,103,104,105,106,107,108,109,110,111,112,113,114,115,116,117,118,119,120],
    "Name": ["Anu","Rahul","Meera","Arun","Priya","Vivek","Sneha","Akhil","Divya","Rohan","Neha","Kiran","Anjali","Manu","Swathi","Aditya","Lakshmi","Nikhil","Pooja","Sanjay"],
    "Age": [25,34,29,45,31,52,27,38,41,24,36,48,30,55,26,43,33,39,28,50],
    "City": ["Kochi","Calicut","Kochi","Bangalore","Chennai","Kochi","Calicut","Chennai","Bangalore","Kochi","Calicut","Kochi","Chennai","Bangalore","Kochi","Calicut","Bangalore","Kochi","Chennai","Calicut"],
    "Account_Type": ["Savings","Current","Savings","Current","Savings","Savings","Current","Savings","Current","Savings","Savings","Current","Savings","Current","Savings","Savings","Current","Savings","Savings","Current"],
    "Balance": [25000,85000,45000,120000,30000,75000,150000,28000,95000,18000,65000,110000,52000,175000,22000,48000,135000,35000,41000,90000],
    "Loan_Amount": [0,500000,200000,800000,0,300000,1000000,0,600000,0,250000,700000,0,900000,0,150000,450000,0,100000,350000],
    "Credit_Score": [780,690,720,650,810,735,620,770,680,820,710,640,790,600,805,750,670,760,795,630],
    "Transactions": [15,32,21,45,18,27,51,14,39,12,29,42,24,55,16,31,47,22,19,36]
}

df = pd.DataFrame(data)
df
```

## Level 1 — Basic Pandas

1. Complete dataset · 2. First five customers · 3. Last five customers · 4. DataFrame shape · 5. Names and cities

```python
print("1. Complete dataset")
display(df)

print("2. First 5 customers")
display(df.head())

print("3. Last 5 customers")
display(df.tail())

print(f"4. Shape: Rows: {df.shape[0]}, Columns: {df.shape[1]}")

print("5. Customer names and cities")
display(df[["Name", "City"]])
```

## Level 2 — Filtering

6. Balance over ₹50,000 · 7. Age under 30 · 8. Kochi customers · 9. Credit score over 750 · 10. Customers with a loan

```python
print("6. Balance greater than ₹50,000")
display(df.loc[df["Balance"] > 50000, ["Name", "Balance"]])

print("7. Customers younger than 30")
display(df.loc[df["Age"] < 30, ["Name", "Age", "City"]])

print("8. Customers from Kochi")
display(df.loc[df["City"] == "Kochi"])

print("9. Credit score greater than 750")
display(df.loc[df["Credit_Score"] > 750, ["Name", "Credit_Score"]])

print("10. Customers who have taken a loan")
display(df.loc[df["Loan_Amount"] > 0, ["Name", "Loan_Amount"]])
```

## Level 3 — Calculations

11. Total bank balance · 12. Average balance · 13. Highest balance · 14. Lowest credit score · 15. Total loans given

```python
print(f"11. Total bank balance: ₹{df['Balance'].sum():,}")
print(f"12. Average account balance: ₹{df['Balance'].mean():,.2f}")

print("13. Customer with the highest balance")
display(df.loc[[df["Balance"].idxmax()], ["Name", "Balance"]])

print("14. Customer with the lowest credit score")
display(df.loc[[df["Credit_Score"].idxmin()], ["Name", "Credit_Score"]])

print(f"15. Total loan amount given by the bank: ₹{df['Loan_Amount'].sum():,}")
```

## Level 4 — Grouping

16. Average balance by city · 17. Customer count by account type · 18. Average credit score by account type

```python
print("16. Average balance by city")
display(df.groupby("City")["Balance"].mean().rename("Average Balance").to_frame())

print("17. Number of customers in each account type")
display(df["Account_Type"].value_counts().rename("Customer Count").to_frame())

print("18. Average credit score by account type")
display(df.groupby("Account_Type")["Credit_Score"].mean().rename("Average Credit Score").to_frame())
```

## Level 5 — Real Banking Analysis

19. Categorize customers based on balance. 20. Mark customers eligible when their credit score is at least 700 **and** their balance is at least ₹30,000.

```python
df["Customer_Category"] = [
    "Premium" if balance >= 100000 else "Regular" if balance >= 50000 else "Basic"
    for balance in df["Balance"]
]
print("19. Customer categories")
display(df[["Name", "Balance", "Customer_Category"]])

df["Loan_Eligible"] = (
    (df["Credit_Score"] >= 700) & (df["Balance"] >= 30000)
).map({True: "Eligible", False: "Not Eligible"})
print("20. Loan eligibility")
display(df[["Name", "Credit_Score", "Balance", "Loan_Eligible"]])
```

## Bonus Challenge

Find the city with the highest average customer balance, then count customers in that city whose credit score is above 700.

```python
average_balance_by_city = df.groupby("City")["Balance"].mean()
top_city = average_balance_by_city.idxmax()
customers_above_700 = df.loc[
    (df["City"] == top_city) & (df["Credit_Score"] > 700)
]

print("Average customer balance by city:")
display(average_balance_by_city.rename("Average Balance").to_frame())
print(f"City with the highest average balance: {top_city}")
print(f"Customers in {top_city} with credit score above 700: {len(customers_above_700)}")
display(customers_above_700[["Name", "Credit_Score", "Balance"]])
```
